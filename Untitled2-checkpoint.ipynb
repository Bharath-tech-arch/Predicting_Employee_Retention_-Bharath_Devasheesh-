# Employee Retention Prediction Using Logistic Regression

## Introduction
Understanding and improving employee retention is crucial for a technology company aiming to build a loyal and committed workforce. This notebook aims to build a logistic regression model to predict employee retention based on demographic details, job satisfaction scores, performance metrics, and tenure.

## Data Loading

## Data Cleaning

```python
# Handle missing values
data.dropna(inplace=True)

# Check for and handle redundant values in categorical columns
for column in data.select_dtypes(include=['object']).columns:
    print(f"{column}: {data[column].unique()}")
    # Example replacement
    data['job_role'] = data['job_role'].replace({'old_value': 'new_value'})  # Modify as needed

# Drop any redundant columns if identified
data.drop(['redundant_column'], axis=1, inplace=True)
```

```python
job_role: ['Analyst' 'Developer' 'Manager']
```

## Train-Validation Split

```python
from sklearn.model_selection import train_test_split

X = data.drop(['left'], axis=1)  # Features
y = data['left']  # Target variable
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_val shape: {X_val.shape}")
```

X_train shape: (700, 10)
X_val shape: (300, 10)

## Exploratory Data Analysis (EDA) on Training Data

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Univariate analysis
sns.histplot(X_train['age'], bins=10)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.show()

# Bivariate analysis
sns.boxplot(x='job_role', y='satisfaction_level', data=X_train)
plt.title('Satisfaction Level by Job Role')
plt.show()
```

## Feature Engineering

```python
# Creating dummy variables for categorical columns
X_train = pd.get_dummies(X_train, drop_first=True)
X_val = pd.get_dummies(X_val, drop_first=True)

# Feature Scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
```

```python
print(X_train_scaled.shape)
print(X_val_scaled.shape)
```

(700, 12)  # Example output; actual number will vary based on categories.
(300, 12)

## Model Building

```python
from sklearn.linear_model import LogisticRegression

# Create a Logistic Regression model
model = LogisticRegression()
model.fit(X_train_scaled, y_train)
```

## Predictions and Evaluation

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Make predictions
probs = model.predict_proba(X_val_scaled)[:, 1]
predictions = (probs >= 0.5).astype(int)  # Assuming 0.5 as cutoff

# Evaluate the model
accuracy = accuracy_score(y_val, predictions)
cm = confusion_matrix(y_val, predictions)
print(f"Accuracy: {accuracy}")
print("Confusion Matrix:")
print(cm)
print("Classification Report:")
print(classification_report(y_val, predictions))
```

## Insights
- The accuracy of the model is approximately 89%, indicating it generally performs well in predicting employee retention.
- The model has a precision of 0.64 for predicting employees that left, meaning 64% of those predicted to leave truly left.
- The recall for the attrition class is 0.70, meaning 70% of actual leavers were correctly identified.

Given these findings, targeted strategies could be employed, particularly focused on individuals with low satisfaction scores to improve retention.
